```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('cleaned_cp_dataset.csv')
```

```python
df.head(10)
```

```python
sample_materials = df['formula'].unique()[:5]

plt.figure(figsize=(9, 6))
for mat in sample_materials:
    subset = df[df['formula'] == mat]
    plt.plot(subset['T'], subset['Cp'], marker='o', label=mat)

plt.title('ael_youngs_modulus_vrh(Cp) vs agl_bulk_modulus_isothermal_300K (T) for Selected Materials')
plt.xlabel('bulkmodulus (K)')
plt.ylabel('youngsmodulus')
plt.legend()
plt.grid(True)
plt.show()
```
