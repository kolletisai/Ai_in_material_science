```python
import matplotlib.pyplot as plt
```

```python


# 1. Define the data for 5 Group 14 Elements
elements = ['C', 'Si', 'Ge', 'Sn', 'Pb']
atomic_numbers = [6, 14, 32, 50, 82]
bandgaps = [5.47, 1.11, 0.67, 0.10, 0.00]

# 2. Initialize the figure size
plt.figure(figsize=(8, 5))

# 3. Create the line plot with markers
plt.plot(atomic_numbers, bandgaps,
         marker='o',
         color='blue',
         linestyle='-',
         linewidth=2)

# 4. Apply Figure Labelling Basics
plt.title('Bandgap vs. Atomic Number (Group 14 Elements)', fontsize=14)
plt.xlabel('Atomic Number (Z)', fontsize=12)
plt.ylabel('Bandgap Energy (eV)', fontsize=12)



# 5. Add a grid and show the plot
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
