```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load dataset
df = pd.read_csv("perovskites_data.csv")

# 2. EDA Visualizations
plt.figure(figsize=(14, 5))

# Plot 1: Bandgap Distribution
plt.subplot(1, 2, 1)
sns.histplot(df["Bandgap_eV"], kde=True, color="teal", bins=30)
plt.title("Perovskite Bandgap Distribution")
plt.xlabel("Bandgap (eV)")
plt.ylabel("Count")

# Plot 2: Formation Energy vs Bandgap (colored by stability)
plt.subplot(1, 2, 2)
scatter = plt.scatter(
    df["Formation_Energy_eV_atom"],
    df["Bandgap_eV"],
    c=df["Energy_Above_Hull_eV"],
    cmap="viridis_r",
    alpha=0.7,
    edgecolors="none",
    s=25,
)
plt.colorbar(scatter, label="Energy Above Hull (eV/atom)")
plt.title("Bandgap vs Formation Energy")
plt.xlabel("Formation Energy (eV/atom)")
plt.ylabel("Bandgap (eV)")

plt.tight_layout()
plt.savefig("perovskites_eda.png", dpi=300)
plt.show()

print("Visualizations created and saved to 'perovskites_eda.png'.")
```
