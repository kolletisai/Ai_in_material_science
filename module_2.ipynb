```python
import matplotlib.pyplot as plt

# Data
metals = ['Aluminum', 'Copper', 'Silver', 'Gold']
thermal_conductivity = [237, 401, 429, 318]

# Create bar chart
plt.figure(figsize=(8, 5))
plt.bar(metals, thermal_conductivity, color='green')

# Labels and title
plt.title('Thermal Conductivity of Common Engineering Metals')
plt.xlabel('Metals')
plt.ylabel('Thermal Conductivity (W/m·K)')

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
