```python
# ============================================================
# Session 10 - ABC3 Perovskite Data Profiling & Imputation
# ============================================================

import pandas as pd
from ydata_profiling import ProfileReport
from sklearn.impute import SimpleImputer, KNNImputer


# ------------------------------------------------------------
# 1. Load dataset from Session 09
# ------------------------------------------------------------

df = pd.read_csv("perovskites_data.csv")

print("✅ Dataset loaded successfully")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

display(df.head(10))


# ------------------------------------------------------------
# 2. Check columns
# ------------------------------------------------------------

print("\n📌 Dataset columns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 3. Check missing values
# ------------------------------------------------------------

print("\n📌 Missing values BEFORE imputation:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 4. Generate profiling report
# ------------------------------------------------------------

profile = ProfileReport(
    df,
    title="ABC3 Perovskites Data Profiling Report",
    explorative=True
)

profile.to_file(
    "perovskites_data_report.html"
)

print("\n✅ Profiling report generated!")
print("📄 perovskites_data_report.html")


# ------------------------------------------------------------
# 5. Mean Imputation
# ------------------------------------------------------------

mean_imputer = SimpleImputer(
    strategy="mean"
)

df_mean_imputed = df.copy()

df_mean_imputed[["Bandgap"]] = (
    mean_imputer.fit_transform(
        df[["Bandgap"]]
    )
)

print("\n✅ Mean imputation completed")


# ------------------------------------------------------------
# 6. KNN Imputation
# ------------------------------------------------------------

numeric_columns = [
    "Bandgap",
    "Formation_Energy_per_Atom",
    "Volume"
]

df_knn_imputed = df.copy()

knn_imputer = KNNImputer(
    n_neighbors=5
)

df_knn_imputed[numeric_columns] = (
    knn_imputer.fit_transform(
        df[numeric_columns]
    )
)

print("✅ KNN imputation completed")


# ------------------------------------------------------------
# 7. Check missing values AFTER imputation
# ------------------------------------------------------------

print("\n📌 Missing values AFTER Mean Imputation:")
print(df_mean_imputed.isnull().sum())

print("\n📌 Missing values AFTER KNN Imputation:")
print(df_knn_imputed.isnull().sum())


# ------------------------------------------------------------
# 8. Save mean-imputed dataset
# ------------------------------------------------------------

df_mean_imputed.to_csv(
    "perovskites_mean_imputed.csv",
    index=False
)

print("\n✅ Mean-imputed dataset saved:")
print("   perovskites_mean_imputed.csv")


# ------------------------------------------------------------
# 9. Save KNN-imputed dataset
# ------------------------------------------------------------

df_knn_imputed.to_csv(
    "perovskites_knn_imputed.csv",
    index=False
)

print("\n✅ KNN-imputed dataset saved:")
print("   perovskites_knn_imputed.csv")


# ------------------------------------------------------------
# 10. Final dataset preview
# ------------------------------------------------------------

print("\n📊 Final KNN-imputed dataset:")

display(
    df_knn_imputed.head(10)
)
```
