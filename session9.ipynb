```python
import pandas as pd
from mp_api.client import MPRester

# Paste your newly generated API key here
API_KEY = "AKfdpjsE4ne2OTlkvuMk770Ap3wl74gF"

with MPRester(API_KEY) as mpr:
    # Query summary data for ABC3 perovskites
    docs = mpr.materials.summary.search(
        formula="ABC3",
        fields=[
            "material_id",
            "formula_pretty",
            "band_gap",
            "formation_energy_per_atom",
            "volume",
            "energy_above_hull",
        ],
    )

# Format into DataFrame
df = pd.DataFrame(
    [
        {
            "Material_ID": str(d.material_id),
            "Formula": d.formula_pretty,
            "Bandgap_eV": d.band_gap,
            "Formation_Energy_eV_atom": d.formation_energy_per_atom,
            "Volume_A3": d.volume,
            "Energy_Above_Hull_eV": d.energy_above_hull,
        }
        for d in docs
    ]
)

# Export results
df.to_csv("perovskites_data.csv", index=False)
print(f"Success! Exported {len(df):,} perovskite entries to 'perovskites_data.csv'.")
```
